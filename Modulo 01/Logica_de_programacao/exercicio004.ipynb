```python
#07 Programa usando for para imprimir somente os numeros inteiros da lista
numeros = [10,5,3,-7,2,-8]
for i in numeros:
    if i < 0:
        break
    print(i)
```

```python
#08 Programa usando for para imprimir todas as palavras da lista, encerrado laço em python
lista = ['java','c++', 'python', 'javascrpt','ruby']
for palavras in lista:
    if palavras == 'python':
        break
    print(palavras)
```

```python
#17/06
#01 Programa que peça dois numeros ao usuario e diga qual é o maior
num1 = int(input('Digite um número '))
num2 = int(input('Digite outro número '))

if num1 > num2:
    print(f'O numero {num1} é maior que o numero {num2}')
elif num1 < num2:
    print(f'O numero {num1} é menor que o numero {num2}')
else:
    print('Número inválido')

```

```python
# 02 Programa que leia três numeros e diga qual é o maior. 
#Faça uso do max e do mim.
num1 = int(input('Digite o primeiro numero '))
num2 = int(input('Digite o segundo numero '))
num3 = int(input('Digite oterceiro numero  '))

print(f'O numero maior é : {max(num1,num2,num3)}\n O numero menor é : {min(num1,num2,num3)}')




```

```python
#04 Programa que peça nota de 0 a 10 e dia se aluno aprovado ou reprovado, aprovado nota maior que 6
nota = float(input('Digite uma nota de 0 a 10:'))
while nota < 0 or nota > 10:
    nota = float(input('Valor atribuido a nota invalido,tente novamente...\nDigite uma nota: '))
if nota >=6:
    print(f'Aprovado nota {nota}')
else: 
    print(f'Reprovado nota {nota}')

```
