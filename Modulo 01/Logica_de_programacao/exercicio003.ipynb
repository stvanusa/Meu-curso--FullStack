```python
#02 Programa para defifir intervala do numero de 0 a 10
num = int(input('Informe um número'))

if num >=1 and num <=10:
    print ('Dentro do intervalo')
else:
    print ('Fora do intervalo')
```

```python
#03 Programa para ler o numero do usuario e informa se é par ou impar
num = int(input('Informe um número'))
if num % 2 == 0 and num > 0:
    print('Par')
else:
    print('Impar')
```

```python
# 04 Programa para classsificar a idade

idade = int(input('Informe sua idade'))

if idade >= 0 and idade <= 12:
    print('Criança')
elif idade >= 13 and idade <=17:
    print('Adolescente')
elif idade >=18 and idade <= 65:
    print('Adulto')
elif idade >=65:
    print('Idoso')
else:
    print('Idade invalida')

```

```python
# 05 Programa para verificar operador matematico
num1 = int(input('Informe o primeiro numero: '))
num2 = int(input('Informe o segundo numero: '))

operacao = input("Escolha a operação (+, -, *, /): ")

if operacao == '+':
    print(num1 + num2)
elif operacao == '-':
    print(num1 - num2)
elif operacao == '*':
    print(num1 * num2)
elif operacao == '/':
    print(num1 / num2) 
else:
    print('Operação invalida')


    
    

    
        
```

```python
# 06 Programa que calcula a média das notas e infoma se reprovado ou aprovado

nota1 = float(input('Informe a primeira '))
nota2 = float(input('Informe a segunda '))
nota3 = float(input('Informe a terceira'))
media = (nota1 + nota2 + nota3)/3

print(f'A média das notas é {media}')

if media >=0 and media <=4:
    print ('reprovado')
elif media >=5 and media <=6:
    print('Recuperação')
elif media >=7 and media <=10:
    print('Aprovado')
else:
    print('Nota invalida')

```
